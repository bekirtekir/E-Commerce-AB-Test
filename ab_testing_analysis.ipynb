```python
import numpy as np
import pandas as pd
from scipy import stats
import math

# 1. Veri Seti Simülasyonu (Örnek: Kontrol ve Test Grupları)
np.random.seed(42)

n_control = 10000  # Kontrol grubundaki ziyaretçi sayısı
n_test = 10000     # Test grubundaki ziyaretçi sayısı

# Kontrol grubu dönüşüm oranı %12, Test grubu dönüşüm oranı %13.5 olsun
conversion_rate_control = 0.12
conversion_rate_test = 0.135

control_conversions = np.random.binomial(1, conversion_rate_control, n_control)
test_conversions = np.random.binomial(1, conversion_rate_test, n_test)

df = pd.DataFrame({
    'group': ['control']*n_control + ['test']*n_test,
    'converted': np.concatenate([control_conversions, test_conversions])
})

# 2. Özet İstatistikler
summary = df.groupby('group')['converted'].agg(['count', 'sum', 'mean']).rename(
    columns={'count': 'Total_Users', 'sum': 'Conversions', 'mean': 'Conversion_Rate'}
)
print("--- Özet İstatistikler ---")
print(summary)

# 3. İstatistiksel Hipotez Testi (Z-Test for Proportions)
from statsmodels.stats.proportion import proportions_ztest

count = [summary.loc['test', 'Conversions'], summary.loc['control', 'Conversions']]
nobs = [summary.loc['test', 'Total_Users'], summary.loc['control', 'Total_Users']]

z_stat, p_value = proportions_ztest(count, nobs)

print("\n--- Hipotez Testi Sonuçları ---")
print(f"Z-statistic: {z_stat:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value < 0.05:
    print("Sonuç: H0 reddedilir! Test grubu ile kontrol grubu arasında istatistiksel olarak anlamlı bir fark var.")
else:
    print("Sonuç: H0 reddedilemez. İki grup arasında anlamlı bir fark bulunamadı.")
```
